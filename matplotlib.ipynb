Matplot Lib 

```python
import matplotlib.pyplot as plt
import numpy as np  
```

1. Line Plot
Use for: Time series, trend analysis.

Function: plt.plot()

```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 40]

plt.plot(x, y, marker='o', color='blue', linestyle='--')
plt.title("Line Plot Example")
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.grid(True)
plt.show()
```

2. Scatter Plot
Use for: Correlation, clustering, pattern analysis.

Function: plt.scatter()

```python
x = np.random.rand(50)
y = np.random.rand(50)

plt.scatter(x, y, color='red')
plt.title("Scatter Plot Example")
plt.xlabel("Feature A")
plt.ylabel("Feature B")
plt.show()
```

3. Bar Chart
Use for: Category comparison.

Function: plt.bar() or plt.barh() (horizontal)

```python
categories = ['Apples', 'Bananas', 'Cherries']
values = [30, 50, 20]

plt.bar(categories, values, color='green')
plt.title("Fruit Sales")
plt.ylabel("Quantity Sold")
plt.show()
```

4. Histogram
Use for: Distribution of numeric data.

Function: plt.hist()

```python
data = np.random.randn(1000)

plt.hist(data, bins=30, color='skyblue', edgecolor='black')
plt.title("Histogram Example")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

5. Pie Chart
Use for: Proportion, percentage distribution.

Function: plt.pie()

```python
labels = ['A', 'B', 'C']
sizes = [40, 35, 25]

plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90)
plt.title("Pie Chart Example")
plt.axis('equal')  # Ensures it's a circle
plt.show()
```

6. Box Plot
Use for: Spread, outliers, distribution shape.

Function: plt.boxplot()

```python
data = [np.random.normal(0, std, 100) for std in range(1, 4)]

plt.boxplot(data)
plt.title("Box Plot Example")
plt.xlabel("Dataset")
plt.ylabel("Values")
plt.show()
```

7. Heatmap (with imshow)
Use for: Correlation matrices, grid heatmaps.

Function: plt.imshow()

```python
matrix = np.random.rand(5, 5)

plt.imshow(matrix, cmap='hot', interpolation='nearest')
plt.colorbar()
plt.title("Heatmap Example")
plt.show()
```

8. Subplots
Use for: Multiple plots in a single figure.

Function: plt.subplot() or plt.subplots()

```python
fig, axs = plt.subplots(1, 2, figsize=(10, 4))

axs[0].plot([1, 2, 3], [1, 4, 9])
axs[0].set_title("Plot 1")

axs[1].bar(['X', 'Y', 'Z'], [5, 7, 3])
axs[1].set_title("Plot 2")

plt.tight_layout()
plt.show()
```
