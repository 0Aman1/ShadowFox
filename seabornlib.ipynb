Seaborn Library
Seaborn works best with DataFrames and integrates tightly with Pandas.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

1. Line Plot
Use for: Trends over continuous data.

Function: sns.lineplot()

```python
df = pd.DataFrame({
    'x': [1, 2, 3, 4, 5],
    'y': [10, 20, 25, 30, 40]
})

sns.lineplot(data=df, x='x', y='y', marker='o')
plt.title("Line Plot (Seaborn)")
plt.show()
```

2. Scatter Plot
Use for: Correlation, clustering.

Function: sns.scatterplot()

```python
df = sns.load_dataset('iris')  # Built-in dataset

sns.scatterplot(data=df, x='sepal_length', y='sepal_width', hue='species')
plt.title("Scatter Plot (Iris Dataset)")
plt.show()
```

3. Bar Chart
Use for: Category-wise aggregation (e.g., mean).

Function: sns.barplot() (aggregates), sns.countplot() (counts)

```python
df = sns.load_dataset('titanic')

sns.barplot(data=df, x='class', y='fare')
plt.title("Average Fare by Class")
plt.show()
```

4. Histogram & KDE
Use for: Distribution plots.

Function: sns.histplot(), sns.kdeplot()

```python
# Histogram
sns.histplot(data=df, x='age', bins=30, kde=True)
plt.title("Age Distribution with KDE")
plt.show()
```

5. Pie Chart (Not directly supported)
❌ Seaborn does not have a native pie chart — use Matplotlib for that.

Alternative (counts via bar):

```python
sns.countplot(data=df, x='sex')
plt.title("Gender Distribution (Bar)")
plt.show()
```

6. Box Plot
Use for: Spread, outliers.

Function: sns.boxplot()

```python
sns.boxplot(data=df, x='class', y='age')
plt.title("Age Distribution by Class")
plt.show()
```

7. Violin Plot
Use for: Distribution + KDE + box plot in one.

Function: sns.violinplot()

```python
sns.violinplot(data=df, x='class', y='age', hue='sex', split=True)
plt.title("Violin Plot - Age by Class and Gender")
plt.show()
```

8. Heatmap
Use for: Correlation matrix or grid visualization.

Function: sns.heatmap()

```python
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Correlation Matrix")
plt.show()
```

9. Pair Plot
Use for: Multi-dimensional exploration.

Function: sns.pairplot()

```python
iris = sns.load_dataset('iris')
sns.pairplot(iris, hue='species')
plt.suptitle("Pair Plot (Iris)", y=1.02)
plt.show()
```
